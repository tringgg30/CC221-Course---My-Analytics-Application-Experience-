```python
# Importing the necessary packages for us to use certain functions.
import pandas as pd
import numpy as np
import qdesc as qd
```

```python
# Loading a data set (SleepData.xlsx) and storing it into a variable.
df = pd.read_excel('SleepData.xlsx')
```

```python
# Check the first 5 rows of the dataframe
df.head()
```

```python
# Check the information of the dataframe
df.info()
```

```python
# Using the desc function from the qdesc package to perform descriptive analysis.
qd.desc(df)
```

```python
# Generate descriptive analysis for sleep hours when grouped by sex.
df.groupby('Sex')['sleep(hrs)/weeknights'].describe()
```

The analysis shows that both males and females sleep an average of about 7.6–7.7 hours on weeknights, with very small differences between the groups. The median sleep is ~8.3 hours for both, indicating most respondents get adequate rest. Variability is similar, though males show slightly higher spread. Overall, sleep patterns are consistent across sexes.



```python
# Generate descriptive analysis for sleep hours when grouped by employment.
df.groupby('Employment')['sleep(hrs)/weeknights'].describe()
```

The results show a clear difference in sleep patterns by employment status. Permanent employees sleep an average of 8.87 hours, with little variation, while probationary employees average only 6.26 hours, with greater variability. This suggests that permanent employees generally get sufficient rest, whereas probationary employees tend to have shorter and less consistent sleep on weeknights.

```python
# Generate descriptive analysis for sleep hours when grouped by marital status.
df.groupby('MaritalStatus')['sleep(hrs)/weeknights'].describe()
```

The results show that married respondents sleep an average of 7.67 hours, while single respondents average 7.58 hours—a very small difference. Both groups have similar variability and medians (~8.3–8.4 hours), suggesting that marital status does not significantly affect weeknight sleep patterns.

```python
# Generate a frequency distribution for a specified variable: Sex
df["Sex"].value_counts()
```

The frequency distribution shows that the dataset has 350 females and 321 males. This indicates a fairly balanced sample, with females making up a slightly larger portion of the respondents.

```python
# Generate a frequency distribution for a specified variable: Employment
df["Employment"].value_counts()
```

The frequency distribution shows that there are 352 permanent employees and 319 probationary employees in the dataset. This indicates a nearly balanced distribution, with a slightly higher number of permanent employees.

```python
# Generate a frequency distribution for a specified variable: Marital Status
df["MaritalStatus"].value_counts()
```

The frequency distribution for marital status shows that the dataset is almost evenly divided, with 338 respondents being single and 333 respondents being married. This indicates a well-balanced representation between the two groups.

```python
# Generate frequency distribution for all the categorical features from the said data frame.
categorical = df.select_dtypes(include=["object", "category"]).columns
for col in categorical:
   counts = df[col].value_counts()
   percentages = df[col].value_counts(normalize=True) * 100
   print(pd.DataFrame({"Counts": counts, "Percentages": percentages}))
   print("\n")


```

The dataset shows a fairly balanced distribution across all categorical variables. In terms of sex, females make up 52.2% while males account for 47.8%, showing only a slight difference. Marital status is almost evenly split, with 50.4% single and 49.6% married respondents. Employment status is also balanced, with 52.5% being permanent employees and 47.5% probationary. Overall, the dataset provides a well-distributed representation across these categories, without any major skew toward a particular group.

```python
# Generating a frequency distribution but this time have it stored in an excel file and organized
# on a per sheet basis per categorical feature.
categorical= df.select_dtypes(include='object').columns
excel_file_path = 'frequency_distributions.xlsx'

with pd.ExcelWriter(excel_file_path) as writer:
    for col in categorical:
        counts = df[col].value_counts()
        percentages = df[col].value_counts(normalize=True) * 100
        freq_df = pd.DataFrame({'Counts': counts, 'Percentages': percentages})
        freq_df.to_excel(writer, sheet_name=col)

print(f"Frequency distributions saved to {excel_file_path}")
```
