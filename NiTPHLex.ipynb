## Performance vs Engagement

```python
# Correlating buffer_events_per_session and content_engagement_score(0-100)
# Choosing the Spearman Rho given that one of the variable is not normal.
spearman_corr, spearman_p = stats.spearmanr(df['buffer_events_per_session'], df['content_engagement_score(0-100)'])
print(f"Spearman correlation coefficient: {spearman_corr:.4f}, p-value: {spearman_p:.4f}")
```

```python
# Checking the scatterplot to visualize the relationship of these variables
sns.scatterplot(x=df['content_engagement_score(0-100)'], y=df['buffer_events_per_session'])
plt.title('Scatter Plot for Linearity Check')
plt.show()
```

```python
# Correlating video_load_time(seconds) and content_engagement_score(0-100)
# Choosing the Spearman Rho given that one of the variable is not normal.
spearman_corr, spearman_p = stats.spearmanr(df['video_load_time(seconds)'], df['content_engagement_score(0-100)'])
print(f"Spearman correlation coefficient: {spearman_corr:.4f}, p-value: {spearman_p:.4f}")
```

```python
# Checking the scatterplot to visualize the relationship of these variables
sns.scatterplot(x=df['video_load_time(seconds)'], y=df['content_engagement_score(0-100)'])
plt.title('Scatter Plot for Linearity Check')
plt.show()
```

```python
# Correlating mobile_error_rate(proportion) and content_engagement_score(0-100)
# Choosing the Spearman Rho given that one of the variable is not normal.
spearman_corr, spearman_p = stats.spearmanr(df['mobile_error_rate(proportion)'], df['content_engagement_score(0-100)'])
print(f"Spearman correlation coefficient: {spearman_corr:.4f}, p-value: {spearman_p:.4f}")
```
